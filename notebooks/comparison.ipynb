```python
import pandas as pd

df_mc = pd.read_csv(r"C:\Users\tedvm\Centrale\PoleProjetS8\Data_PublicHealth_CentraleSupelec\data\media_cloud\mediacloud_health_FR_monthly_2015_2026_wide.csv")
df_gt = pd.read_csv(r"C:\Users\tedvm\Centrale\PoleProjetS8\Data_PublicHealth_CentraleSupelec\data\google_trends\google_trends_health_FR_monthly_2015_2026_wide.csv")

display(df_mc.head())
display(df_gt.head())
```

```python
#normalisation des données media cloud
for col in df_mc.columns:
    if col != "date":
        df_mc[col] = (df_mc[col]/df_mc[col].max())*100
```

```python
import matplotlib.pyplot as plt

for col in df_gt.columns:
    plt.figure(figsize=(12, 4))
    plt.plot(df_gt.index, df_gt[col], label="google trends")
    plt.plot(df_mc.index, df_mc[col], label="media cloud")
    plt.xlabel("Date")
    plt.ylabel("Valeur")
    plt.title("Comparaison de deux séries temporelles")
    plt.legend()
    plt.show()
```
